```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .master("spark://iabd-spark-master:7077")
    .appName("spark-mongodb")

    # --- Classpath ---
    .config("spark.driver.extraClassPath", "/opt/spark/extra-jars/*")
    .config("spark.executor.extraClassPath", "/opt/spark/extra-jars/*")

    .getOrCreate()
)
```

```python
url_iabd = "jdbc:mysql://iabd-mysql:3306/retail_db?useSSL=false&allowPublicKeyRetrieval=true"
propiedades = {
    "driver": "com.mysql.cj.jdbc.Driver",
    "user": "iabd",
    "password": "iabd"
}
```

```python
df = spark.read.jdbc(url=url_iabd,\
    table="customers",\
    properties=propiedades)
```
