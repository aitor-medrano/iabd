# Proyecto Series Temporales


## 1. Cargando los datos

```python
import pandas as pd

# Cargamos los datos
df_temperatura = pd.read_csv("ts_temp_20170201_20200131.csv")
df_energia = pd.read_csv("ts_load_metered_20170201_20200131.csv")
```

```python
df_temperatura.head()
```

```python
# Pasamos el campo DATE a datetime
df_temperatura['DATE'] = pd.to_datetime(df_temperatura['DATE'])
```

```python
df_energia.head()
```

```python
# Pasamos el campo datetime_beginning_ept a datetime
df_energia['datetime_beginning_ept'] = pd.to_datetime(df_temperatura['DATE'])
```

```python
# Unimos los datos por fecha
df_merged = pd.merge(df_energia, df_temperatura, left_on='datetime_beginning_ept', right_on='DATE')
df_merged.head()
```

```python
df = df_merged[["DATE", "mw", "HourlyDryBulbTemperature"]]
df.columns = ["fecha", "mw", "temperatura"]

df.head()
```

```python
# Otra forma
columnas = {'fecha': df_temperatura['DATE'], 'mw': df_energia['mw'], 'temp': df_temperatura['HourlyDryBulbTemperature'], }
```

```python
df = pd.DataFrame(columnas)
df.head()
```

## 2. Columnas con campos hora, dia, semana, mes y año

```python
df['hora'] = pd.DatetimeIndex(df['fecha']).hour
df['dia'] = pd.DatetimeIndex(df['fecha']).day
df['diasemana'] = pd.DatetimeIndex(df['fecha']).dayofweek
df['mes'] = pd.DatetimeIndex(df['fecha']).month
df['anyo'] = pd.DatetimeIndex(df['fecha']).year

df.head()
```

```python
df = df.set_index('fecha')
df.head()
```

## 3. Comprueba las temperaturas y rellena los huecos

```python
# Comprobamos si hay valores nulos en la columna de la temperatura
sum(df['temp'].isnull())
```

```python
df['temp'] = df['temp'].interpolate(method='linear')
```

```python
# Comprobamos de nuevo
sum(df['temp'].isnull())
```

## 4. Separamos el dataset

```python
train = pd.DataFrame(df['2017-01-01':'2019-12-31'])
test = pd.DataFrame(df['2020-01-01':'2020-01-31'])
```
