```python
import datetime
import dlt
from dlt.sources.sql_database import sql_table
```

```python
pedidos = sql_table(
    credentials="mysql+pymysql://root:rootpass@mysql-retail:3306/retail_db",
    table="orders",
    incremental=dlt.sources.incremental("order_date", initial_value=datetime.datetime(2014, 1, 1)),
    write_disposition="merge",
    primary_key="order_id",
)
```

```python
pipeline = dlt.pipeline(
    pipeline_name="retail_el",
    destination="duckdb",
    dataset_name="raw",
)
```

```python
info = pipeline.run(pedidos, refresh="drop_resources")
print(info)
print("filas cargadas:", pipeline.last_trace.last_normalize_info.row_counts.get("orders", 0))
print("paquetes de carga:", len(info.load_packages))
```
