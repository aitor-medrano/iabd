```python
from audiomentations import Compose, AddGaussianNoise, TimeStretch, PitchShift, Shift
import librosa

augment = Compose([
    AddGaussianNoise(min_amplitude=0.001, max_amplitude=0.015, p=0.5), #ruido
    TimeStretch(min_rate=0.9, max_rate=1.25, p=0.5), # velocidad
    PitchShift(min_semitones=-4, max_semitones=4, p=0.5) # tono
])
```

```python
y, sr = librosa.load(librosa.example('nutcracker'))
```

```python
print(f"Duración: {len(y) / sr}")
```

```python
from IPython.display import Audio
Audio(data=y, rate=sr)
```

```python
import matplotlib.pyplot as plt
import librosa.display

plt.figure().set_figwidth(12)
librosa.display.waveshow(y, sr=sr)
```

```python
# Quitamos silencios delante/detrás
audio_trimmed, index = librosa.effects.trim(y)
```

```python
print(f"Duración: {len(audio_trimmed) / sr}")
```

```python
import matplotlib.pyplot as plt
import librosa.display

plt.figure().set_figwidth(12)
librosa.display.waveshow(audio_trimmed, sr=sr)
```

```python
augmented_samples = augment(samples=y, sample_rate=sr)
```

```python
print(f"Duración: {len(augmented_samples) / sr}")
```

```python
import matplotlib.pyplot as plt
import librosa.display

plt.figure().set_figwidth(12)
librosa.display.waveshow(augmented_samples, sr=sr)
```

```python
from IPython.display import Audio
Audio(data=augmented_samples, rate=sr)
```
